```python
def dfs(graph,node,visited):
  if node not in visited:
    print(node,end="")
    visited.add(node)
    for neighbour in graph [node]:
      dfs(graph,neighbour,visited)

graph={}
n=int(input("enter the number of vertices:"))
for i in range(n):
  vertex = input(f"enter vertex {i+1}:")
  neighbours = input(f"Enter adjacent vertices of {vertex} (space-separated):").split()
  graph[vertex]=neighbours

start = input("Enter the starting vertex:")
print("\nGraph:",graph)
print("DFS traversail:",end="")

visited = set()
dfs(graph, start,  visited)
```

```python
from collections import deque
def bfs(graph,start):
  visited = set()
  queue = deque([start])
  visited.add(start)
  while queue:
    vertex = queue.popleft()
    print(vertex,end="")
    for neighbour in graph[vertex]:
      if neighbour not in visited:
        visited.add(neighbour)
        queue.append(neighbour)

graph={}
n = int(input("Enter the number of vertices:"))
for i in range (1,n+1):
  vertex = input(f"Enter vertex {i}:")
  neighbours = input(f"Enter neighbours of {vertex}(space-separated):").split()
  graph[vertex] = neighbours

start = input ("Enter the starting vertex:")
bfs(graph, start)
```
