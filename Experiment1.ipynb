```python
from collections import deque
def bfs(graph,start):
  visited = set()
  queue = deque([start])
  visited.add(start)
  print("BFS Traversal:",end="")
  while queue:
    vertex = queue.popleft()
    print(vertex,end="")
    if vertex in graph: # Check if the vertex exists as a key in the graph
      for neighbour in graph[vertex]:
        if neighbour not in visited:
          visited.add(neighbour)
          queue.append(neighbour)

graph = {}
n = int(input("Enter the number of number of nodes:"))

# Populate the graph with all nodes, initially with empty neighbor lists
# This ensures all potential nodes are keys in the graph dictionary.
for i in range(n):
    vertex = input(f"Enter node {i+1}:")
    graph[vertex] = []

# Now, ask for connections (edges)
for vertex in list(graph.keys()): # Iterate through existing nodes
  neughbours_input = input(f"Enter adjacent nodes of {vertex} (space-separated, leave blank if none):")
  if neughbours_input:
    neighbours = neughbours_input.split()
    # Ensure all entered neighbours are also nodes in the graph
    for nbr in neighbours:
      if nbr not in graph:
        graph[nbr] = [] # Add new nodes if they are mentioned as neighbors
    graph[vertex].extend(neighbours)

start = input("Enter the starting vertex for BFS:")

# Validate starting vertex
if start not in graph:
  print(f"Error: Starting vertex '{start}' not found in the graph.")
else:
  print("\nGraph:",graph)
  bfs(graph,start)
```
